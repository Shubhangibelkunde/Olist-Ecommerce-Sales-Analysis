# Step 3 - Handling Missing value

3A - orders missing value treatment

```python
# STEP 3A — ORDERS TABLE: Handle Nulls
# -----------------------------------------------

# Check all order statuses first
print("Order Status Counts:")
print(orders['order_status'].value_counts())
print(f"\nTotal orders: {len(orders)}")

# Keep ONLY delivered orders (nulls in date columns won't matter after this)
orders_delivered = orders[orders['order_status'] == 'delivered'].copy()

print(f"\nDelivered orders kept: {len(orders_delivered)}")
print(f"Removed (cancelled/other): {len(orders) - len(orders_delivered)}")

# Verify — nulls should be very low now
print("\nNulls after filtering delivered only:")
print(orders_delivered.isnull().sum())
```

```python
# Drop the 24 rows where delivery dates are still null
# These are delivered orders with missing timestamps — unusable for time analysis

orders_delivered = orders_delivered.dropna(subset=[
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date'
])

print(f"Final clean orders: {len(orders_delivered)}")
print("\nNull check after drop:")
print(orders_delivered.isnull().sum())
```

## 3B — PRODUCTS TABLE
# Columns with nulls: product_category_name, product_name_lenght, etc.

```python
print("\nProducts nulls:")
print(products.isnull().sum())
 
# Fill missing category names with 'unknown'
products['product_category_name'] = products['product_category_name'].fillna('unknown')
 
# Fill numeric nulls with median (safe approach)
num_cols = ['product_name_lenght', 'product_description_lenght',
            'product_photos_qty', 'product_weight_g',
            'product_length_cm', 'product_height_cm', 'product_width_cm']
 
for col in num_cols:
    products[col] = products[col].fillna(products[col].median())
 
print("✅ Products nulls handled!")
```

```python
# 3C — REVIEWS TABLE
# review_comment_title and review_comment_message can be null (customer didn't write)
# ACTION: Fill with empty string — not needed for analysis
```

```python
reviews['review_comment_title']   = reviews['review_comment_title'].fillna('')
reviews['review_comment_message'] = reviews['review_comment_message'].fillna('')
 
print("✅ Reviews nulls handled!")
```

 STEP 4 — FIX DATA TYPES (DATES)
# -------------------------------------------------------
# All date columns are loaded as 'object' (string) — convert to datetime

```python
date_cols_orders = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
 
for col in date_cols_orders:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')
    orders_delivered[col] = pd.to_datetime(orders_delivered[col], errors='coerce')
print("✅ All date columns converted to datetime!")
print(orders.dtypes)
```

```python
# Reviews dates
reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'], errors='coerce')
reviews['review_answer_timestamp'] = pd.to_datetime(reviews['review_answer_timestamp'], errors='coerce')
 
print("✅ All date columns converted to datetime!")
```

 STEP 5 — REMOVE DUPLICATES

```python
print(f"\nDuplicates in orders: {orders.duplicated().sum()}")
print(f"Duplicates in order_items: {order_items.duplicated().sum()}")
print(f"Duplicates in customers: {customers.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
print(f"Duplicates in reviews: {reviews.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
print(f"Duplicates in payments: {payments.duplicated().sum()}")
 
# Remove if any found
orders      = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
customers   = customers.drop_duplicates()
payments    = payments.drop_duplicates()
reviews     = reviews.drop_duplicates(subset='review_id')  # keep first review per order
 
print("✅ Duplicates removed!")
```

```python
# Check duplicates in ALL 9 tables at once

tables = {
    'orders': orders,
    'order_items': order_items,
    'customers': customers,
    'products': products,
    'sellers': sellers,
    'payments': payments,
    'reviews': reviews,
    'geolocation': geolocation,
    'category_trans': category_trans
}

print("Duplicate Check — All 9 Tables:")
print("-" * 40)
for name, df in tables.items():
    dupes = df.duplicated().sum()
    status = "⚠️ HAS DUPLICATES" if dupes > 0 else "✅ Clean"
    print(f"{name:<20} {dupes:>6} duplicates   {status}")
```

```python
# STEP 1 — See how many duplicates
print(f"Total geolocation rows: {len(geolocation)}")
print(f"Duplicate rows: {geolocation.duplicated().sum()}")

# STEP 2 — See what the duplicates look like
print("\nSample duplicate rows:")
print(geolocation[geolocation.duplicated(keep=False)].sort_values('geolocation_zip_code_prefix').head(10))

# STEP 3 — Fix: Keep one row per zip code (first occurrence)
geolocation_clean = geolocation.drop_duplicates(subset='geolocation_zip_code_prefix', keep='first')

print(f"\nBefore: {len(geolocation)} rows")
print(f"After:  {len(geolocation_clean)} rows")
print(f"Removed: {len(geolocation) - len(geolocation_clean)} duplicates ✅")
```

STEP 6 — FEATURE ENGINEERING (Create New Useful Columns)

```python
# 6A — Delivery time in days
orders_delivered['delivery_days'] = (
    orders_delivered['order_delivered_customer_date'] -
    orders_delivered['order_purchase_timestamp']
).dt.days
```

```python
# 6B — Was delivery late? (delivered after estimated date)
orders_delivered['is_late'] = (
    orders_delivered['order_delivered_customer_date'] >
    orders_delivered['order_estimated_delivery_date']
).astype(int)   # 1 = late, 0 = on time
```

```python
# 6C — Extract year, month from purchase date (for trend analysis)
orders_delivered['purchase_year']  = orders_delivered['order_purchase_timestamp'].dt.year
orders_delivered['purchase_month'] = orders_delivered['order_purchase_timestamp'].dt.month
orders_delivered['purchase_month_name'] = orders_delivered['order_purchase_timestamp'].dt.strftime('%b')
orders_delivered['purchase_yearmonth'] = orders_delivered['order_purchase_timestamp'].dt.to_period('M').astype(str)
 
print("✅ Feature engineering done!")
print(orders_delivered[['order_id','delivery_days','is_late','purchase_yearmonth']].head())
```

STEP 7 — TRANSLATE CATEGORY NAMES (Portuguese → English)

```python
# Merge translation table into products
products = products.merge(category_trans, on='product_category_name', how='left')
products['product_category_name_english'] = products['product_category_name_english'].fillna('unknown')
 
print("✅ Category names translated to English!")
print(products[['product_category_name','product_category_name_english']].head())
```

# STEP 8 — MERGE ALL TABLES INTO ONE MASTER TABLE

```python
# Start with delivered orders
master = orders_delivered.copy()
 
# Add order items (price, freight)
master = master.merge(order_items, on='order_id', how='left')
 
# Add customer info
master = master.merge(customers, on='customer_id', how='left')
 
# Add product info
master = master.merge(products[['product_id','product_category_name_english',
                                 'product_weight_g']], on='product_id', how='left')
 
# Add seller info
master = master.merge(sellers[['seller_id','seller_city','seller_state']],
                       on='seller_id', how='left')
 
# Add review scores
reviews_clean = reviews[['order_id','review_score']].drop_duplicates(subset='order_id')
master = master.merge(reviews_clean, on='order_id', how='left')
 
# Add payment info (total payment per order)
payment_summary = payments.groupby('order_id').agg(
    total_payment=('payment_value', 'sum'),
    payment_type=('payment_type', 'first')
).reset_index()
master = master.merge(payment_summary, on='order_id', how='left')
 
print(f"\n✅ Master table created: {master.shape}")
print(master.columns.tolist())
```

STEP 9 — FINAL CHECKS & CLEAN UP

```python
# Check remaining nulls
print("\nFinal null check:")
print(master.isnull().sum()[master.isnull().sum() > 0])
 
# Remove rows where price is null or zero (bad records)
master = master[master['price'] > 0]
 
# Remove extreme outliers in price (above 99th percentile — likely data errors)
price_99 = master['price'].quantile(0.99)
master = master[master['price'] <= price_99]
print(f"\nAfter outlier removal: {master.shape}")
 
# Reset index
master = master.reset_index(drop=True)
```

```python
# FIX 1 — review_score: 1360 nulls
# Some customers never left a review — fill with median score
master['review_score'] = master['review_score'].fillna(master['review_score'].median())
print(f"Review score median: {master['review_score'].median()}")

# FIX 2 — total_payment & payment_type: only 3 nulls — safe to drop
master = master.dropna(subset=['total_payment', 'payment_type'])

print(f"\nRows after fixing: {len(master)}")

# FINAL NULL CHECK
print("\nFinal null check — should be 0:")
remaining = master.isnull().sum()
print(remaining[remaining > 0] if remaining[remaining > 0].any() else "✅ Zero nulls — Master table is 100% clean!")

# QUICK SUMMARY
print(f"\n{'='*45}")
print(f"✅ MASTER TABLE READY")
print(f"   Rows    : {master.shape[0]:,}")
print(f"   Columns : {master.shape[1]}")
print(f"   Date range: {master['order_purchase_timestamp'].min().date()} → {master['order_purchase_timestamp'].max().date()}")
print(f"   Avg order value : ₹{master['total_payment'].mean():,.2f}")
print(f"   Avg review score: {master['review_score'].mean():.2f} / 5")
print(f"{'='*45}")
```
